seperate missingness into category , numeric and date

```python
numeric_cols = df.select_dtypes(
    include=['int64', 'float64']
).columns

categorical_cols = df.select_dtypes(
    include=['object']
).columns
```

```python
for col in numeric_cols:
  if df[col].isnull().sum()>0:
    df[col]=df[col].fillna(df[col].median())
```

```python
print(numeric_cols.isnull().sum())
```

```python
for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna('Unknown')
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

Encoding

```python
x=df.drop(columns=['Risk_Score','Loan_Status'])
y_classification=df['Loan_Status']
y_regression=df['Risk_Score']
```

```python
categorical_columns = x.select_dtypes(include='object').columns

print(categorical_columns.tolist())
```

```python
x_encoded=pd.get_dummies(x,columns=categorical_columns,drop_first=True).astype(int)
print(
    x_encoded.shape
)
```
