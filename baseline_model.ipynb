```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train_scaled =scaler.fit_transform(x_train)
x_val_scaled = scaler.transform(x_val)
x_test_scaled = scaler.transform(x_test)
```

```python
y_train_reg = pd.to_numeric(y_train_reg)
y_val_reg = pd.to_numeric(y_val_reg)
y_test_reg = pd.to_numeric(y_test_reg)
```

```python
print(y_train_reg.dtype)
print(y_val_reg.dtype)
print(y_test_reg.dtype)
```

```python

from sklearn.linear_model import LinearRegression
baseline_reg=LinearRegression()
```

```python
baseline_reg.fit(x_train_scaled,y_train_reg)
```

```python
y_pred_reg=baseline_reg.predict(x_val_scaled)
```

```python
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(
    y_val_reg,
    y_pred_reg
)

print("MAE:", mae)
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

rmse = np.sqrt(
    mean_squared_error(
        y_val_reg,
        y_pred_reg
    )
)

print("RMSE:", rmse)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(
    y_val_reg,
    y_pred_reg
)

print("R²:", r2)
```

```python
print("===== LINEAR REGRESSION BASELINE =====")
print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")
```
