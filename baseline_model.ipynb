```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression
from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
X = pd.read_csv(
    '/content/drive/MyDrive/X_encoded_preprocessed.csv'
)

y_classification = pd.read_csv(
    '/content/drive/MyDrive/y_classification.csv'
).squeeze()

y_regression = pd.read_csv(
    '/content/drive/MyDrive/y_regression.csv'
).squeeze()
```

```python
print(X.head())
```

```python
print(y_classification.head())
print(y_regression.head())
```

```python
print("X shape:", X.shape)
print("Classification target:", y_classification.shape)
print("Regression target:", y_regression.shape)
```

```python
derived_features = [
    'Income_Stability_Index',
    'Farm_Productivity_Index',
    'Climate_Risk_Index',
    'Financial_Health_Index',
    'Loan_Affordability_Index',
    'Asset_Strength_Index',
    'Overall_Risk_Factor'
]

print(X[derived_features].head())
```

```python
X = X.drop(columns=derived_features)

print("New X shape:", X.shape)
```

```python
print([col for col in derived_features if col in X.columns])
```

for classification

```python
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X,
    y_classification,
    test_size=0.20,
    random_state=42,
    stratify=y_classification
)
```

for regression

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X,
    y_regression,
    test_size=0.20,
    random_state=42

)
```

```python
print("Classification:")
print(X_train_cls.shape)
print(X_test_cls.shape)

print("\nRegression:")
print(X_train_reg.shape)
print(X_test_reg.shape)
```

scaling for classification

```python
from sklearn.preprocessing import StandardScaler
Scalar=StandardScaler()
X_train_cls=Scalar.fit_transform(X_train_cls)
X_test_cls=Scalar.transform(X_test_cls)
Scalar.fit_transform(X_test_cls)

```

scaling for regression model

```python
from sklearn.preprocessing import StandardScaler
Scalar=StandardScaler()
X_train_reg=Scalar.fit_transform(X_train_reg)
X_test_reg=Scalar.transform(X_test_reg)
Scalar.fit_transform(X_test_reg)
```

logistic regression for loan status

```python
baseline_classifer=LogisticRegression(
    random_state=42,
    max_iter=700
)
baseline_classifer.fit(X_train_cls,y_train_cls)
```

```python
y_pred_cls=baseline_classifer.predict(X_test_cls)
```

```python
accuracy = accuracy_score(y_test_cls,y_pred_cls)
print(f" Baseline Accuracy: { accuracy:.4f}")
print(f" Baseline Accuracy: { accuracy*100:.2f}")
```

```python
print(classification_report(
    y_test_cls,
    y_pred_cls
))
```

now training baseline model for risk score

```python
from sklearn.linear_model import LinearRegression
baseline_model=LinearRegression()
baseline_model.fit(X_train_reg,y_train_reg)
```

```python
y_pred_reg=baseline_model.predict(X_test_reg)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test_reg, y_pred_reg)

rmse = np.sqrt(
    mean_squared_error(y_test_reg, y_pred_reg)
)

r2 = r2_score(y_test_reg, y_pred_reg)

print("Baseline Regression Results")
print("---------------------------")
print("MAE :", mae*100)
print("RMSE:", rmse*100)
print("R²  :", r2*100)
```
