outliers

```python
outlier_cols = [
    'Farm_Size_Acres',
    'Annual_Farm_Income',
    'Annual_Farm_Expenses',
    'Asset_Value',
    'Liability_Value',
    'Loan_Amount_Requested',
    'Existing_Loan_Amount'
]

for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: {count} outliers")
```

```python
outlier_data = []
for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()
    percentage = (count / len(df)) * 100
    outlier_data.append({
        'Column_Name': col,
        'Outlier_Count': count,
        'Outlier_Percentage': percentage
    })
outlier_report = pd.DataFrame(outlier_data)
outlier_report.sort_values(
    'Outlier_Percentage',
    ascending=False
)
```
