```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('/content/drive/MyDrive/agrishield_ai_loan_dataset.csv')
df.shape
```

```python

duplicate_columns = [
    col for col in df.columns
    if col != 'Farmer_ID'
]

print(
    "Duplicate application profiles:",
    df.duplicated(subset=duplicate_columns).sum()
)

df = df.drop_duplicates(
    subset=duplicate_columns,
    keep='first'
).reset_index(drop=True)

print("Shape after duplicate removal:", df.shape)
```

```python
print("No. of columns :",len(df.columns))
print("No. of rows :",len(df))
```

splitting the data set into x and y features

```python
x=df.drop(columns=['Risk_Score','Loan_Status'])
y_regression=df['Risk_Score']
y_classification=df['Loan_Status']
```

```python
print(y_regression.head())
print(y_classification.head())
```

splitting the dataset into 70% training , 30 temporary (15 validation ,15 test)

```python
from sklearn.model_selection import train_test_split
x_train , x_temp , y_train_reg , y_temp_reg , y_train_cls , y_temp_cls= train_test_split(x,
                                                                                            y_regression,
                                                  y_classification,

                                                  test_size=0.3,random_state=42,
                                                  stratify=y_classification)

```

spliting the 30% into validation and testing

```python
x_val, x_test, y_val_reg, y_test_reg, y_val_cls, y_test_cls = train_test_split(
    x_temp,
    y_temp_reg, # This contains the numerical Risk_Score for the temporary set
    y_temp_cls, # This contains the categorical Loan_Status for the temporary set
    test_size=0.50,
    random_state=42,
    stratify=y_temp_cls # Correct: Stratify on the classification target (Loan_Status)
)
```

```python
print("Shape of x_train:", x_train.shape)
print("Shape of x_val:", x_val.shape)
print("Shape of x_test:", x_test.shape)
print("Shape of y_train_reg:", y_train_reg.shape)
print("Shape of y_val_reg:", y_val_reg.shape)
print("Shape of y_test_reg:", y_test_reg.shape)
print("Shape of y_train_cls:", y_train_cls.shape)
print("Shape of y_val_cls:", y_val_cls.shape)
print("Shape of y_test_cls:", y_test_cls.shape)
```

handling missing values

```python
numerical_cols=x_train.select_dtypes(include=['int64','float64']).columns.tolist()
category_cols=x_train.select_dtypes(include=['object']).columns.tolist()
```

```python
for col in numerical_cols:
  median_value=x_train[col].median()
  x_train[col]=x_train[col].fillna(median_value)
  x_val[col]=x_val[col].fillna(median_value)
  x_test[col]= x_test[col].fillna(median_value)
```

```python
for col in category_cols:

    x_train[col] = x_train[col].fillna('Unknown')
    x_val[col] = x_val[col].fillna('Unknown')
    x_test[col] = x_test[col].fillna('Unknown')
```

```python
print("Train missing:", x_train.isna().sum().sum())
print("Validation missing:", x_val.isna().sum().sum())
print("Test missing:", x_test.isna().sum().sum())
```

clean/remove columns before encoding

```python
x_train=x_train.drop(columns='Farmer_ID')
x_val=x_val.drop(columns='Farmer_ID')
x_test=x_test.drop(columns='Farmer_ID')
```

feature extration from application date into year , month , day , quater or month , day of week

```python
for data in [x_train, x_val, x_test]:

    data['Application_Date'] = pd.to_datetime(
        data['Application_Date'],
        errors='coerce'
    )
```

```python
for data in [x_train, x_val, x_test]:

    data['Application_Year'] = data['Application_Date'].dt.year
    data['Application_Month'] = data['Application_Date'].dt.month
    data['Application_Day'] = data['Application_Date'].dt.day
    data['Application_DayOfWeek'] = data['Application_Date'].dt.dayofweek
    data['Application_Quarter'] = data['Application_Date'].dt.quarter
```

```python
print(x_train[
    [
        'Application_Date',
        'Application_Year',
        'Application_Month',
        'Application_Day',
        'Application_DayOfWeek',
        'Application_Quarter'
    ]
].head())
```

```python
x_train = x_train.drop(columns=['Application_Date'])
x_val = x_val.drop(columns=['Application_Date'])
x_test = x_test.drop(columns=['Application_Date'])
```

```python
numerical_cols = x_train.select_dtypes(include=['int64', 'float64']).columns.tolist()
categorical_cols = x_train.select_dtypes(include=['object']).columns.tolist()

print("Updated Numerical columns:", len(numerical_cols))
print("Updated Categorical columns:", len(category_cols))
print("\nUpdated Categorical columns list:")
print(categorical_cols)
```

```python
derived_features = [
    'Income_Stability_Index',
    'Farm_Productivity_Index',
    'Climate_Risk_Index',
    'Financial_Health_Index',
    'Loan_Affordability_Index',
    'Asset_Strength_Index',
    'Overall_Risk_Factor'
]
```

```python
for col in derived_features:
  print(col, "->" , col in x_train.columns)
```

encoding


```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(handle_unknown='ignore',sparse_output=False)
```

```python
# Convert categorical columns to string type to ensure uniformity for the encoder
x_train[categorical_cols] = x_train[categorical_cols].astype(str)
x_val[categorical_cols] = x_val[categorical_cols].astype(str)
x_test[categorical_cols] = x_test[categorical_cols].astype(str)

x_train_encoded = encoder.fit_transform(x_train[categorical_cols])
x_val_encoded = encoder.transform(x_val[categorical_cols])
x_test_encoded = encoder.transform(x_test[categorical_cols])

# Get feature names for the encoded columns
encoded_feature_names = encoder.get_feature_names_out(categorical_cols)

print("Shape of x_train_encoded:", x_train_encoded.shape)
print("Shape of x_val_encoded:", x_val_encoded.shape)
print("Shape of x_test_encoded:", x_test_encoded.shape)
```

convert the arrays into dataframe

```python
encoded_feature_names = encoder.get_feature_names_out(
    categorical_cols
)

x_train_encoded = pd.DataFrame(
    x_train_encoded,
    columns=encoded_feature_names,
    index=x_train.index
)

x_val_encoded = pd.DataFrame(
    x_val_encoded,
    columns=encoded_feature_names,
    index=x_val.index
)

x_test_encoded = pd.DataFrame(
    x_test_encoded,
    columns=encoded_feature_names,
    index=x_test.index
)
```

```python
x_train_final = pd.concat(
    [
        x_train[numerical_cols],
        x_train_encoded
    ],
    axis=1
)

x_val_final = pd.concat(
    [
        x_val[numerical_cols],
        x_val_encoded
    ],
    axis=1
)

x_test_final = pd.concat(
    [
        x_test[numerical_cols],
        x_test_encoded
    ],
    axis=1
)
```

```python
print("Train:", x_train_final.shape)
print("Validation:", x_val_final.shape)
print("Test:", x_test_final.shape)

print("\nMissing values:")
print("Train:", x_train_final.isnull().sum().sum())
print("Validation:", x_val_final.isnull().sum().sum())
print("Test:", x_test_final.isnull().sum().sum())
```
