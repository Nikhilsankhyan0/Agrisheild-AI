encoding

```python
from sklearn.preprocessing import OneHotEncoder

categorical_cols = x_train_cls.select_dtypes(
    include='object'
).columns

# Convert all categorical columns to string type before encoding
for col in categorical_cols:
    x_train_cls[col] = x_train_cls[col].astype(str)
    x_val_cls[col] = x_val_cls[col].astype(str)
    x_test_cls[col] = x_test_cls[col].astype(str)

encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

x_train_encoded = encoder.fit_transform(
    x_train_cls[categorical_cols]
)

x_val_encoded = encoder.transform(
    x_val_cls[categorical_cols]
)

x_test_encoded = encoder.transform(
    x_test_cls[categorical_cols]
)
```

```python
print("Categorical columns:", len(categorical_cols))
print("Encoded training shape:", x_train_encoded.shape)
```

```python
x_val_encoded=encoder.transform(x_val_cls[categorical_cols])
x_test_encoded=encoder.transform(x_test_cls[categorical_cols])
```
