**FEATURE AUDIT**

What is this column, what type of information does it contain, and what role does it play in AgriShield AI

```python
feature_audit=pd.DataFrame({
    "Column_Name":df.columns,
    "Data_Type":df.dtypes.astype(str).values,
    'Unique_Values':df.nunique().values,
    'Missing_Values':df.isna().sum().values,
})
print(feature_audit)
```

```python
feature_audit['Missing_value_percentage']=(
    df.isnull().mean().values*100
).round(2)
print(feature_audit[["Column_Name","Missing_value_percentage"]])
```

**feature Classifiation**

classify features into categories like identifier , date , Demographocical , Agricultural , Environmental , Financial , Credit , loan , Derived Features and target

```python
feature_audit.loc[feature_audit['Column_Name']=='Farmer_ID','Business_Role']="identifier"
feature_audit.loc[feature_audit['Column_Name']=='Application_Date','Business_Role']="date"
```

```python
demographic_features = [
    'Age',
    'Gender',
    'Education_Level',
    'Marital_Status',
    'Family_Size',
    'Dependents',
    'Farming_Experience_Years'
]

feature_audit.loc[
    feature_audit['Column_Name'].isin(demographic_features),
    'Business_Role'
] = 'Demographic'
```

```python
Geographical_Feature=['State','District']
feature_audit.loc[feature_audit['Column_Name'].isin(Geographical_Feature),'Business_Role']='Geographical'
```

```python
agricultural_features = [
    'Farm_Size_Acres',
    'Land_Ownership',
    'Soil_Type',
    'Soil_Quality_Index',
    'Irrigation_Availability',
    'Primary_Crop',
    'Secondary_Crop',
    'Average_Yield_Last3Years_TonsPerAcre',
    'Crop_Insurance',
    'Government_Subsidy',
    'MSP_Eligible',
    'Tractor_Ownership',
    'Farm_Equipment_Value',
    'Livestock_Count'
]
feature_audit.loc[feature_audit['Column_Name'].isin(agricultural_features),'Business_Role']='Agricultural'
```

```python
environmental_features = [
    'Annual_Rainfall_mm',
    'Average_Temperature_C',
    'Humidity_Percent',
    'Flood_Risk',
    'Drought_Risk',
    'Cyclone_Risk',
    'Pest_Risk',
    'Weather_Risk_Index',
    'Climate_Zone'
]
feature_audit.loc[feature_audit['Column_Name'].isin(environmental_features),'Business_Role']='environmental_feature'
```

```python
financial_features = [
    'Annual_Farm_Income',
    'Annual_Farm_Expenses',
    'Asset_Value',
    'Liability_Value',
    'Savings',
    'Bank_Balance'
]
feature_audit.loc[feature_audit['Column_Name'].isin(financial_features),'Business_Role']='finacial_features'
```

```python
credit_features = [
    'Credit_Score',
    'Existing_Loan_Count',
    'Existing_Loan_Amount',
    'Previous_Default',
    'Repayment_History_Score',
    'Debt_to_Income_Ratio'
]
feature_audit.loc[feature_audit['Column_Name'].isin(credit_features),'Business_Role']='credit_features'
```

```python
loan_features = [
    'Loan_Amount_Requested',
    'Loan_Purpose',
    'Interest_Rate_Percent',
    'Loan_Tenure_Months',
    'EMI'
]
feature_audit.loc[feature_audit['Column_Name'].isin(loan_features),'Business_Role']='loan_features'
```

```python
derived_features = [
    'Income_Stability_Index',
    'Farm_Productivity_Index',
    'Climate_Risk_Index',
    'Financial_Health_Index',
    'Loan_Affordability_Index',
    'Asset_Strength_Index',
    'Overall_Risk_Factor'
]
feature_audit.loc[feature_audit['Column_Name'].isin(derived_features),'Business_Role']='derived_features'
```

```python
target_features = [
    'Risk_Score',
    'Loan_Status'
]
feature_audit.loc[feature_audit['Column_Name'].isin(target_features),'Business_Role']='target_features'
```

```python
identifier_feature = [id_col]
date_feature = ['Application_Date']
geographic_features = Geographical_Feature

feature_groups = {
    'Identifier': identifier_feature,
    'Date': date_feature,
    'Demographic': demographic_features,
    'Geographic': geographic_features,
    'Agricultural': agricultural_features,
    'Environmental': environmental_features,
    'Financial': financial_features,
    'Credit': credit_features,
    'Loan': loan_features,
    'Derived Feature': derived_features,
    'Target': target_features
}
```

```python
pd.set_option('display.max_rows', None)
for role, columns in feature_groups.items():
    feature_audit.loc[
        feature_audit['Column_Name'].isin(columns),
        'Business_Role'
    ] = role
```

**DATA QUALITY**

```python
missing_report = pd.DataFrame({
    'Missing_Count': df.isnull().sum(),
    'Missing_Percentage': (df.isnull().mean() * 100).round(2)
})

missing_report = missing_report.sort_values(
    'Missing_Percentage',
    ascending=False
)
missing_report
```

**Which 5 columns have the highest missing percentage?**

Crop_Insurance	5.94
Cyclone_Risk		5.93
Marital_Status	5.74
Soil_Quality_Index	5.68
Average_Yield_Last3Years_TonsPerAcre  5.61

**Which columns have 0 missing values?**

Farmer id ,Appilcation_date , Risk_Score, Loan_Status



```python
missing_report[missing_report['Missing_Count'] > 3]['Missing_Percentage'].plot(
    kind='bar',
    figsize=(21, 10)
)

plt.title('Missing Values by Feature')
plt.xlabel('Features')
plt.ylabel('Missing Percentage')
plt.xticks(rotation=90)
plt.show()
```
