As debt burden increases, does the proportion of rejected applications change?(DTI--Debt to income ratio)

```python
df['DTI_Band'] = pd.cut(
    df['Debt_to_Income_Ratio'],
    bins=[-np.inf, 0.2, 0.4, 0.6, 1.0, np.inf],
    labels=[
        'Very Low',
        'Low',
        'Moderate',
        'High',
        'Very High'
    ]
)
pd.crosstab(
    df['DTI_Band'],
    df['Loan_Status'],
    normalize='index'
).mul(100).round(2)
```

Risk of loan default based on the weather condition
(Does higher environmental risk correspond to different loan outcomes?)

```python
df['Weather_Risk_Band'] = pd.qcut(
    df['Weather_Risk_Index'],
    q=4,
    labels=[
        'Low',
        'Medium-Low',
        'Medium-High',
        'High'
    ]
)
pd.crosstab(
    df['Weather_Risk_Band'],
    df['Loan_Status'],
    normalize='index'
).mul(100).round(2)
```

**MULTIVARIATE ANALYSIS**

When multiple variables are considered together, do they tell a consistent story about credit risk?

```python
import matplotlib.pyplot as plt
import seaborn as sns
top_risk_features = ['Credit_Score',
                     'Financial_Health_Index',
                     'Repayment_History_Score',
                     'Existing_Loan_Amount',
                     'Asset_Strength_Index',
                     'Overall_Risk_Factor',
                     'Weather_Risk_Index',
                     'Debt_to_Income_Ratio',
                     'Risk_Score'
                     ]
corr_risk=df[top_risk_features].corr()
plt.figure(figsize=(10,5))
sns.heatmap(
    corr_risk,
    annot=True,
    fmt='.1f',
    cmap='coolwarm',
    center=0
)

plt.title('Correlation Among Key Risk Features')
plt.show()
```

```python
import seaborn as sns

derived_correlations = df[derived_features].corr()
plt.figure(figsize=(10, 4))
sns.heatmap(
    derived_correlations,
    annot=True,
    fmt='.1f',
    cmap='viridis',
    center=0
)
plt.title('Correlation Among Derived Features')
plt.show()
```
